```python
def highest_score(*args):
    highest = args[0]

    for score in args:
        if score > highest:
            highest = score

    return highest


result = highest_score(0.2, 0.8, 0.5, 0.9, 0.6)
print("Highest score:", result)
```

```python
def calculate_average(*args):
    total = sum(args)
    count = len(args)
    average = total / count
    return average


model1 = calculate_average(0.80, 0.85, 0.90)
model2 = calculate_average(0.88, 0.82, 0.91)

print("Model 1 average:", round(model1, 2))
print("Model 2 average:", round(model2, 2))

if model1 > model2:
    print("Model 1 performs better")
elif model2 > model1:
    print("Model 2 performs better")
else:
    print("Both models perform equally")
    
```

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result
    return wrapper


@my_decorator
def calculate_accuracy(correct, total):
    accuracy = (correct / total) * 100
    return accuracy


result = calculate_accuracy(90, 100)
print("Model accuracy:", result, "%")
```

```python
# Lambda function to calculate squared error
squared_error = lambda actual, predicted: (actual - predicted) ** 2

# Regular function to calculate total squared error
def total_error():
    error1 = squared_error(10, 8)
    error2 = squared_error(20, 18)
    error3 = squared_error(30, 27)

    total = error1 + error2 + error3
    return total

print("Total squared error:", total_error())
```

```python
# 1. Create a decorator
def training_decorator(func):
    def wrapper(*args, **kwargs):
        print("Training started")
        result = func(*args, **kwargs)
        print("Training finished")
        return result
    return wrapper


# 2. Create a lambda function to reduce loss
reduce_loss = lambda loss, amount: round(loss - amount, 2)


# 3. Create the training function
@training_decorator
def train_model(starting_loss, reduction_amount, epochs):
    loss = starting_loss

    for i in range(1, epochs + 1):
        loss = reduce_loss(loss, reduction_amount)

        print("Epoch:", i, "Loss:", f"{loss:.2f}")

        if loss <= 0.10:
            print("Stopping training")
            break

    return loss


# 4. Call the function
final_loss = train_model(1.0, 0.05, 20)

print("Final loss:", f"{final_loss:.2f}")
```
